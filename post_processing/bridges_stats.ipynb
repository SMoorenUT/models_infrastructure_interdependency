```python
import json
import pandas as pd
import os
from pathlib import Path

CURR_DIR = Path(os.getcwd())

def main():
    # read 
    with open(CURR_DIR.parent / 'data/init_data/bridges.json', 'r') as file:
        bridges_data = json.load(file)

    # read file
    current_directory = Path(os.getcwd())
    file_path = current_directory.parent / 'output_simulations/ema_road_model_08_05_2024/bridges/ema_road_model_08_05_2024_transport.volume_to_capacity_ratio_2050.csv'
    data = pd.read_csv(file_path, index_col=0)
    # data.index = bridges_data['data']['bridge_entities']['reference']
    # calculate mean of each row
    data['mean_IC_2050'] = data.mean(axis=1)

    # sort dataframe descending on mean
    data_sorted = data.sort_values(by='mean_IC_2050', ascending=False)
    data_sorted.head()

    # save to csv
    data_sorted.to_csv(CURR_DIR.parent / 'output_simulations/ema_road_model_08_05_2024/bridges/ema_road_model_08_05_2024_transport.volume_to_capacity_ratio_2050_sorted.csv')

    return data_sorted

print(main())
```
